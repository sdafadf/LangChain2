```python
from pydantic import Field

from typing_extensions import TypedDict


#理解问题
class MovieDict(TypedDict):
    title:str
    year:int
    director:str
    rating:float
    
movie : MovieDict= {
    "title1":"盗梦空间",
    "year1":2021,
    "director":"克里斯托弗.诺兰",
    "rating1":8.8,
}
print(movie)
```
