```python
from langchain.chat_models import init_chat_model

#传入模型实例
model=init_chat_model(
    model="deepseek-v4-flash",
    model_provider="deepseek", 
    extra_body={"thinking":{"type":"disabled"}}
)
agent=create_agent(
    model=model
)

print(type(agent))
```

```python
#如何调用agent
from  rich import print  as  rprint
agent=create_agent(
    model=model,
)
response=agent.invoke({"messages":[
    {"role":"user","content":"你好"}
]})
rprint(response)
```

```python
from langchain_core.tools import tool


#如何调用agent
@tool(parse_docstring=True)
def get_weather(city:str):
   """描述天气系统
   
   Args:
       city:具体的城市
   """
   return f"{city}天气晴朗，温度15度"
   
from  rich import print  as  rprint
agent=create_agent(
    model=model,
    tools=[get_weather]
)
response=agent.invoke({"messages":[
    {"role":"system","content":"你是一个查询天气的助手,根据用户的提问查询天气。如果问题和天气无关你可以说我不清楚这个问题"},
    {"role":"user","content":"北京的天气如何"}
]})
rprint(response)
```

```python
from langchain_tavily import TavilySearch

#调用langchain的内置的工具from langchain_core.tools import tool

load_dotenv()
#如何调用agent
web_search=TavilySearch(
    max_results=2,
    # tavily_api_key=os.getenv("TAVILY_API_KEY")
)   
from  rich import print  as  rprint
agent=create_agent(
    model=model,
    tools=[web_search],
    name="agent01"
)
response=agent.invoke({"messages":[
    {"role":"user","content":"帮我查询一下东阳的天气"}
]})
rprint(response)
```

```python
#系统提示词的设置
# 举例一
from langchain_tavily import TavilySearch

#调用langchain的内置的工具from langchain_core.tools import tool

load_dotenv()
#如何调用agent
web_search=TavilySearch(
    max_results=2,
    # tavily_api_key=os.getenv("TAVILY_API_KEY")
)   
from  rich import print  as  rprint
agent=create_agent(
    model=model,
    tools=[web_search],
    system_prompt="你是一个多才多艺的智能智能助手，可以调用工具帮用户解决问题"
)
response=agent.invoke({"messages":[
    {"role":"user","content":"帮我查询一下浙江广厦建设职业技术大学水平咋样"}
]})
rprint(response)
```

```python
from pydantic import BaseModel, Field
from langchain.agents.structured_output import ProviderStrategy, ToolStrategy, AutoStrategy
from rich import  print as rprint

class ContractInfo(BaseModel):
    """用户的联系方式"""
    name:str=Field(description="用户的姓名")
    email:str=Field(description="用户的邮箱")
    phone:str=Field(description="用户的电话")
agent=create_agent(
    model=model,
    response_format=AutoStrategy(ContractInfo)
)
response=agent.invoke({
    "messages": [
        {"role":"user","content":"从下列信息提取用户信息，小明的邮箱是shksa@qq.com,电话是17858003714"}
    ]
})
rprint(response)
```
