```python
import os

from langchain.chat_models import init_chat_model
#流式调用
from dotenv import load_dotenv
load_dotenv()
llm=init_chat_model(
    model="deepseek-v4-flash",
    model_provider="deepseek",
)
```

```python
for chunk in llm.stream("帮我解释一下什么是人工智能"):
    print(chunk.text,end="",flush=True)
```

```python
#2.批量调用（batch)
#举例1：一次性接收所有响应
message=[
    "你好,你是谁？","2+3=?","where is the capital of china"
]
responses=llm.batch(message)
for response in  responses:
    print(response)
```

```python
#按照完成顺序接收响应
#2.批量调用（batch)
#举例1：一次性接收所有响应
message=[
    "你好,你是谁？","2+3=?","where is the capital of china"
]
responses=llm.batch_as_completed(message)
for response in  responses:
    print(response)
```

```python
# 准备多个输入
inputs = [
    "翻译成英文：春天来了",
    "翻译成英文：夏天很热",
    "翻译成英文：秋天落叶",
    "翻译成英文：冬天下雪"
]

# ✅ 批量调用（高效）
import time

start = time.time()
responses = llm.batch(inputs)
batch_time = time.time() - start

print("批量调用结果：")
for i, response in enumerate(responses):
    print(f"{i+1}. {response.content}")
print(f"耗时: {batch_time:.2f}秒\n")
```

```python
#演示顺序调用
#invoke
# ❌ 循环调用（低效，仅用于对比）
inputs = [
    "翻译成英文：春天来了",
    "翻译成英文：夏天很热",
    "翻译成英文：秋天落叶",
    "翻译成英文：冬天下雪"
]

start = time.time()
loop_responses = []
for inp in inputs:
    response = llm.invoke(inp)
    loop_responses.append(response)
loop_time = time.time() - start

for i, response in enumerate(responses):
    print(f"{i+1}. {response.content}")

print(f"循环调用耗时: {loop_time:.2f}秒")

print(f"批量调用节省: {((loop_time - batch_time) / loop_time * 100):.1f}%")
```
