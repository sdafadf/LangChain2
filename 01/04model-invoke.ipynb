```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
load_dotenv()
message=[
    {"role":"system","content":"你是一个专业的数学老师"},{"role":"user","content":"帮我解释一下斐波那契数列"}
]
model=init_chat_model(
    model="deepseek-v4-flash",
    model_name="deepseek",
)
response=model.invoke(message)
print(response)
```

```python
conversation=[
    {"role":"system","content":"你是一个非常友好的ai助手"},
    {"role":"user","content":"你好我叫小明"}
]
print(model.invoke(conversation))
conversation.append({"role":"assistant","content":model.invoke(conversation).content})
conversation.append({"role":"user","content":"我叫什么名字"})
print(model.invoke(conversation))
```

```python
from langchain_core.messages import HumanMessage, SystemMessage

#1.3消息对象列表的输入
conversation=[
    
    SystemMessage(content="你是一个非常友好的ai助手"),
    HumanMessage(content="你好我叫小明")
]
print(model.invoke(conversation))
```

```python
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage

#1.3消息对象列表的输入
conversation=[
    
    SystemMessage(content="你是一个非常友好的ai助手"),
    HumanMessage(content="你好我叫小明")
]
print(model.invoke(conversation))
conversation.append(AIMessage(content=model.invoke(conversation)))
conversation.append(HumanMessage(content="我刚刚问了什么问题"))
print(model.invoke(conversation))
```

```python
#2.invoke返回值
from rich import print as rprint
# 美化返回 
response=model.invoke([HumanMessage(content="2+3=?")])
rprint(response)
```

```python
#举例二
#遇到相关要求
response = model.invoke("用一句话解释什么是 AI")

# 1. 获取回复内容
print("AI 回复:", response.content)

# 2. 获取响应元数据
metadata = response.response_metadata
print(f"使用的模型: {metadata['model_name']}")
print(f"结束原因: {metadata['finish_reason']}")
print(f"模型提供商：{metadata['model_provider']}\n")

# 3. 获取 Token 使用情况
usage = metadata.get('token_usage', {})
print(f"输入 tokens: {usage.get('prompt_tokens')}")
print(f"输出 tokens: {usage.get('completion_tokens')}")
print(f"总计 tokens: {usage.get('total_tokens')}")

# 4. 获取消息 ID
print(f"消息 ID: {response.id}")
```
