```python
import dotenv
from langchain_core.prompts import ChatPromptTemplate, HumanMessagePromptTemplate

# ChatPromptTemplate的使用
# 1.1ChatPromptTemplate的实列化
chat_prompt=ChatPromptTemplate.from_messages([
    ("system","你是一个友好的ai,名字叫{name};"),
    ("human","你最近好吗"),
    ("ai","我很好谢谢"),
    ("human","{user_input}")
])
result=chat_prompt.invoke({"name":"小智","user_input":"1+2=?"})
print(result)
```

```python
#chatPromptTemplate调用方法
# invoke 参数类型 promptvalue ,format 字符串 ,format_messages,数据列表
```

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
load_dotenv()
llm=init_chat_model(
    model="deepseek-v4-flash",
    model_provider="deepseek"
)
chat_prompt=ChatPromptTemplate.from_messages([
    ("system","你是一个友好的ai,名字叫{name};"),
    ("human","你最近好吗"),
    ("ai","我很好谢谢"),
    ("human","{user_input}")
])
result=chat_prompt.invoke({"name":"小智","user_input":"1+2=?"})
response=llm.invoke(result)
print(response)
```

```python
from langchain_core.messages import SystemMessage, HumanMessage
#消息列表不能用变量
chat_prompt=ChatPromptTemplate.from_messages([
   SystemMessage(content="你是一个友好的ai"),
   HumanMessage(content="你好，我叫{name}")
])
chat_prompt.invoke({"name":"小明"})
```

```python
from langchain_core.messages import SystemMessage, HumanMessage
from langchain_core.prompts import SystemMessagePromptTemplate,HumanMessagePromptTemplate
system_mess_template=SystemMessagePromptTemplate.from_template("你是一个友好的ai")
human_mess_template=HumanMessagePromptTemplate.from_template("你好，我叫{name}")
#消息列表不能用变量
chat_prompt=ChatPromptTemplate.from_messages([
system_mess_template,human_mess_template
])
chat_prompt.invoke({"name":"小明"})
```
