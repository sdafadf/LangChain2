```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os
# 补全消息类导入
from langchain_core.messages import AIMessage, HumanMessage, ToolMessage
from langchain_core.tools import tool

# 从.env文件中加载环境变量
load_dotenv(override=True)

# 初始化DeepSeek模型
model = init_chat_model(
    model="deepseek-v4-flash",
    model_provider="deepseek",
)

# 标准LangChain工具装饰器定义函数
@tool
def get_weather(city: str) -> str:
    """查询指定城市天气
    Args:
        city: 需要查询天气的城市名称
    """
    return f"{city}天气不错哦~今日晴朗"

# 绑定工具到模型
model_with_tools = model.bind_tools([get_weather])

# 1. 用户提问
human_msg = HumanMessage(content="北京天气如何")

# 2. 让模型自动生成带tool_calls的AIMessage（不要手动硬编码）
ai_msg = model_with_tools.invoke([human_msg])

# 关键：DeepSeek思考模式必须带上 reasoning_content
# 给AI消息补充reasoning_content字段，规避400报错
if hasattr(ai_msg, "additional_kwargs"):
    ai_msg.additional_kwargs["reasoning_content"] = ""

# 3. 构造工具返回消息，tool_call_id和模型生成的id保持一致
tool_call_id = ai_msg.tool_calls[0]["id"]
tool_msg = ToolMessage(
    content="今天北京天气晴朗，万里无云~",
    tool_call_id=tool_call_id,
    name="get_weather"
)

# 组装完整消息链
messages = [human_msg, ai_msg, tool_msg]

# 最终调用模型整合工具结果
response = model_with_tools.invoke(messages)
print(response.content)
```

```python
#多人对话聊天机器人
load_dotenv()
llm=init_chat_model(
    model="deepseek-v4-flash",
    model_provider="deepseek",
)
Exit_Word="quit"
i=1
messages=[
    SystemMessage("你是一个阿黄大哥客服，耐心的智能客服")
    
]
print(f"请输入具体的问题，输入{Exit_Word}结束对话")
while True:
    print("\n","="*10,f"第{i}轮对话开始","="*10,"\n")
    user_input=input("请输入：")
    if user_input==Exit_Word:
        print("会话结束")
        break
    human_msg=HumanMessage(content=user_input)
    messages.append(human_msg)
    response=llm.invoke(messages)
    print(response.content)
    messages.append(response)
    i+=1
```
